```python
import numpy as np
import pandas as pd


def create_dataset():
    """Create and save a synthetic house-price dataset."""
    rng = np.random.default_rng(42)

    number_of_houses = 60

    area = rng.integers(800, 3001, number_of_houses)
    bedrooms = rng.integers(2, 6, number_of_houses)
    bathrooms = rng.integers(1, 4, number_of_houses)
    age = rng.integers(0, 31, number_of_houses)
    distance = rng.uniform(1, 20, number_of_houses).round(1)

    noise = rng.normal(0, 20, number_of_houses)

    price = (
        area * 0.12
        + bedrooms * 18
        + bathrooms * 25
        - age * 3.5
        - distance * 4
        + 50
        + noise
    ).round(1)

    data = pd.DataFrame(
        {
            "Area_sqft": area,
            "Bedrooms": bedrooms,
            "Bathrooms": bathrooms,
            "Age_years": age,
            "Distance_km": distance,
            "Price_lakh": price,
        }
    )

    data.to_csv("house_prices.csv", index=False)

    print("Dataset created successfully.")
    print(f"Rows: {data.shape[0]}")
    print(f"Columns: {data.shape[1]}")
    print()
    print(data.head())


if __name__ == "__main__":
    create_dataset()
```
